# Stage 9: Seller Table Generation (Task 3 Daily Feeds)

This stage consolidates all 15 daily seller CSV files from `../../data_given/task_3/seller_data/` into a unified `seller_table`.
- Adds `Last time collected` using the filename date (e.g., `2025-12-21`).
- Standardizes column names and orders as defined in the Data Dictionary.
- Handles missing `Screenshot` and `Live_link` entries.
- Exports the consolidated table to `../../data_code/derived/seller_table.csv`.

```python
# 1. Ingest and concatenate all daily seller CSV files
from pathlib import Path

seller_data_folder = Path("../../data_given/task_3/seller_data")
seller_files = sorted(seller_data_folder.glob("*.csv"))

print(f"Found {len(seller_files)} daily seller feed files.")

seller_dfs = []
for file in seller_files:
    df = pd.read_csv(file)
    # Extract date from filename (e.g. '2025-12-21')
    df["Last time collected"] = file.stem
    seller_dfs.append(df)

seller_table_raw = pd.concat(seller_dfs, ignore_index=True)
print(f"Concatenated {len(seller_table_raw)} total records.")
seller_table_raw.head()
```

```python
# 2. Standardize column names and ordering
seller_table = seller_table_raw.rename(
    columns={
        "region": "Region",
        "adv_price": "Advertised_price",
        "live_link": "Live_link",
    }
)

# Ensure Screenshot and Live_link columns exist
if "Screenshot" not in seller_table.columns:
    seller_table["Screenshot"] = None
if "Live_link" not in seller_table.columns:
    seller_table["Live_link"] = None

seller_table["Screenshot"] = seller_table["Screenshot"].fillna("")
seller_table["Live_link"] = seller_table["Live_link"].fillna("")

# Order columns according to the Data Dictionary specification
columns_order = [
    "Seller_name",
    "SKU",
    "Region",
    "Advertised_price",
    "Marketplace",
    "Screenshot",
    "Live_link",
    "Last time collected",
]

seller_table = seller_table[columns_order]
seller_table.head()
```

```python
# 3. Quality checks and summary
print("Shape of Seller Table:", seller_table.shape)
print("\nMissing values per column:\n", seller_table.isna().sum())
print(f"\nUnique collection dates: {seller_table['Last time collected'].nunique()}")
print("\nRecords per collection date:\n", seller_table["Last time collected"].value_counts().sort_index())
```

```python
# 4. Export consolidated Seller Table to CSV
seller_table_output = Path("../../data_code/derived/seller_table.csv")
seller_table.to_csv(seller_table_output, index=False)
print(f"Seller Table exported successfully to {seller_table_output} with {len(seller_table)} rows.")
```
